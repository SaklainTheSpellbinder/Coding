**Statistics**

```python
stats=np.array([[1,2,19],[9,5,6]])
print(stats)
```

```python
print(np.min(stats)) # overall ke min sheta ber kore dey
print(np.min(stats,axis=0)) # proti column e ke min sheta ber kore dey
print(np.min(stats,axis=1))#proti row te ke min sheta ber kore dey
```

same kahini gulo max er khetreo hoy

```python
#sum
print(np.sum(stats)) # overall ke ssum ber kore dey
print(np.sum(stats,axis=0)) # proti column er sum ber kore dey
print(np.sum(stats,axis=1))#proti row er sum ber kore dey
```

**Reorganizing Arrays**

```python
before=np.array([[1,2,3,4],[5,6,7,8]])
print(before)

after=before.reshape((4,2))
print(after)

after2=before.reshape((2,2,2))
print(after2)
```

```python
#Vertical stack
v1=np.array([1,2,3,4])
v2=np.array([5,6,7,8])

np.vstack([v1,v2,v1,v2,v1,v1,v1])
```

```python
#Horizontal stack
h1=np.ones((2,3))
h2=np.zeros((2,2))
np.hstack([h1,h2,h1,h1])
```

**Data from the txt**

```python
hehe=np.genfromtxt('data.txt',delimiter=',')
print(hehe)
hehe2=hehe.astype('int32')
print(hehe2)
```

**Boolean Masking and Advanced Indexing**

```python
print(np.any(hehe2>50))
print(np.any(hehe2>50,axis=0))
print(np.any(hehe2>50,axis=1)) #any hocche kono jekono ekta 50 r beshi holei puratar jonno true print hobe
print()
print(np.all(hehe2>50))
print(np.all(hehe2>50,axis=0))
print(np.all(hehe2>50,axis=1)) 
```
